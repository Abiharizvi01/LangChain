# 1. What is LangChain?
**LangChain is a framework/library for building applications powered by Large Language Models (LLMs).**



# 2. What problem does LangChain solve?


The main problem: Building a real LLM application requires connecting the LLM with many other things, such as your data, databases, APIs, tools, prompts, and multiple steps.

Doing all of this manually can become complicated and require a lot of repetitive code.

LangChain provides ready-made components and a structure to connect these things easily.

For example, suppose you want to build a chatbot that answers questions from your PDFs.

Without LangChain, you would have to manually handle:

Reading PDFs
Splitting text
Finding relevant information
Sending that information to the LLM
Constructing prompts
Handling the response
LangChain provides components to help with these tasks.

In one sentence:
LangChain solves the problem of managing and connecting the different components needed to build an LLM-powered application.

It doesn't make the LLM itself smarter—it makes building the application around the LLM easier and more organized.



# Components of LangChain

LangChain is mainly divided into **6 important components**:

1. Models
2. Prompts
3. Chains
4. Memory
5. Indexes
6. Agents

### Quick revision

1. Models   → AI models
2. Prompts  → Instructions for models
3. Chains   → Connect multiple steps
4. Memory   → Maintain conversation/context
5. Indexes  → Work with external data
6. Agents   → Decide and use tools/actions

# Prompts

A **prompt** is the instruction/input given to the model.

LangChain provides tools to create **reusable and structured prompts**.

**Purpose:** Control what we ask the model and how we ask it.

## 1. Dynamic & Reusable Prompts

A **PromptTemplate** allows us to create a prompt with placeholders (variables) that can be filled with different values.

## 2. Role-Based Prompts

In chat-based LLMs, we can define different **roles** such as:

- **System** → Defines the behavior or role of the AI.
- **User** → Contains the user's request.
- **Assistant** → Represents the AI's response.

## 3. Few-Shot Prompting

**Few-shot prompting** means providing the LLM with a few examples of the expected input and output before giving it a new input.

# Memory

LLM api calls are stateless. let suppose i am using chatgpt api and we asked one question say who is narendra modi it will give answer in the next call we ask how old is he? so in this call he will not have the context from the previous call so he will struggle to decode "he" here.So this problem is solved by this component of langchain.


LLMs are generally **stateless**, meaning they do not automatically remember previous interactions. Memory allows an application to store and provide previous information to the LLM when needed.

For example:

**User:** My name is Ali.  
**AI:** Nice to meet you, Ali.  
**User:** How old is he?

Without the previous conversation being provided, the LLM may not know who **"he"** refers to.

LangChain provides different approaches to manage conversation memory:

### 1. ConversationBufferMemory
Stores the **complete transcript of recent messages**.

- Good for short conversations.
- Can become large quickly as the conversation grows.

### 2. ConversationBufferWindowMemory
Stores only the **last N interactions/messages**.

- Helps reduce token usage.
- Useful when we don't need the entire conversation history.

### 3. Summarizer-Based Memory
Periodically **summarizes older parts of the conversation** instead of storing the complete conversation.

- Keeps a condensed version of previous conversations.
- Helps reduce the amount of information that needs to be sent to the LLM.

### 4. Custom Memory
Allows us to create our own memory mechanism for specific use cases.

For example, we can store specialized information such as:

- User preferences
- Important facts about the user
- Application-specific state

**Purpose of Memory:**  
To maintain relevant context from previous interactions so that the LLM can respond appropriately in an ongoing conversation.

# Agents

An **AI Agent** is an AI system that can **perform a task by deciding what steps to take and which tools to use**.

An AI Agent mainly has two important capabilities:

### 1. Reasoning Capability
The agent can **analyze the user's request, break it into steps, and decide what needs to be done**.

### 2. Tool-Using Capability
The agent can **use external tools** such as APIs, databases, calculators, search engines, Python functions, etc., to perform actions or get information.

### Example

User asks:

> "Find the weather in Delhi and tell me whether I should carry an umbrella."

The agent can reason:

```text
1. I need the current weather in Delhi.
2. I should use a weather API.
3. Check the chance of rain.
4. Based on the result, decide whether an umbrella is needed.
5. Give the user the answer.
```

Here:

- **Reasoning** → deciding these steps.
- **Tool** → weather API used to get the actual weather data.

### In short:

> **AI Agent = Reasoning + Tools → Ability to autonomously perform tasks.**
